```python
from sklearn.cluster import MiniBatchKMeans
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
X,_=make_blobs(n_samples=1500,centers=5,n_features=2,random_state=48)
mbk_means=MiniBatchKMeans(n_clusters=5,batch_size=1500,random_state=48)
mbk_means.fit(X)
plt.scatter(X[:,0],X[:,1],c=mbk_means.labels_)
plt.scatter(mbk_means.cluster_centers_[:,0],mbk_means.cluster_centers_[:,1],s=200,c='red',marker='X')
plt.title('MiniBatchKMeans')
plt.show()
```
